## 1. Anatomy of a command
`ls -lh course/` has three parts: the **command** (`ls`), a **flag** (`-lh`, how) and an **argument** (`course/`, what to act on). Spaces separate the parts and case matters.

The installation guide already used `pwd`, `mkdir`, `cd`, `wget`, `unzip` and `ls -lth`. Now we learn what they mean.

````
pwd
ls
ls -lh
ls --help | head -n 5
````

**Expected output** (example):
```
$ pwd
/home/your_username
$ ls
course
miniforge3
$ ls -lh
total 8.0K
drwxr-xr-x 2 your_username your_username 4.0K Nov 18 11:52 course
drwxr-xr-x 2 your_username your_username 4.0K Nov 18 11:52 miniforge3
$ ls --help | head -n 5
Usage: ls [OPTION]... [FILE]...
List information about the FILEs (the current directory by default).
Sort entries alphabetically if none of -cftuvSUX nor --sort is specified.

Mandatory arguments to long options are mandatory for short options too.
```

`pwd` prints where you are, `ls` lists what is here, `-l` gives a long list and `-h` makes sizes human-readable. `--help` explains almost any command.

## 2. The filesystem and moving around
```
/
├── home
│   └── your_username      # ~ is here
│       └── course
├── mnt
│   └── c                  # your Windows C: drive
└── bin
```
- **Absolute path** starts at `/`, e.g. `/home/your_username/course`
- **Relative path** starts from where you are, e.g. `../course`
- `.` here, `..` parent, `~` home, `/` root

**WSL tip:** work in your Linux home (`~`), not `/mnt/c`. It is faster and avoids permission surprises.

**Predict first:** what will `pwd` show after `cd practice`?

````
pwd
ls
mkdir practice
cd practice
pwd
cd ..
cd -
cd ~
pwd
````

**Expected output** (example):
```
$ pwd
/home/your_username
$ ls
course
miniforge3
$ mkdir practice
$ cd practice
$ pwd
/home/your_username/practice
$ cd ..
$ cd -
/home/your_username/practice
$ cd ~
$ pwd
/home/your_username
```

`cd -` jumps back to the previous folder. Typos are normal: read the error message, press ↑ and fix it.

Nested folders in one go:

````
mkdir -p practice/a/b/c
ls practice
````

**Expected output** (example):
```
$ mkdir -p practice/a/b/c
$ ls practice

```

## 6. Copy, move, remove
`cp` keeps the original, `mv` moves or renames, and **`rm` is permanent: no trash, no undo.** Run `ls` first to see what you will delete.

````
mkdir -p case_1
cp outbreak.fasta case_1/
mv case_1/outbreak.fasta case_1/copy.fasta
ls case_1/
rm case_1/copy.fasta
ls case_1/
````

**Expected output** (example):
```
$ mkdir -p case_1
$ cp outbreak.fasta case_1/
$ mv case_1/outbreak.fasta case_1/copy.fasta
$ ls case_1/
copy.fasta
$ rm case_1/copy.fasta
$ ls case_1/
```

## 7. Redirection: save the output
- `>` write to a file (**overwrites** without asking)
- `>>` add to the end of a file
- `|` send the output to the next command

````
cat wiv04.fasta outbreak.fasta > all_seqs.fasta
grep -c ">" all_seqs.fasta
echo "my notes" > notes.txt
echo "more notes" >> notes.txt
cat notes.txt
````

**Expected output** (example):
```
$ cat wiv04.fasta outbreak.fasta > all_seqs.fasta
$ grep -c ">" all_seqs.fasta
13
$ echo "my notes" > notes.txt
$ echo "more notes" >> notes.txt
$ cat notes.txt
my notes
more notes
```

`all_seqs.fasta` is the file we will align with MAFFT on Day 2.

## 8. Build a pipe one step at a time
Add one command, check the result, then add the next. Use `head` while testing.

````
grep ">" all_seqs.fasta | head -n 3
grep ">" all_seqs.fasta | sed 's/>//' | head -n 3
grep ">" all_seqs.fasta | sed 's/>//' > headers.txt
wc -l headers.txt
````

## 9. Working with tables (TSV)
Find the column number first, then use it. Your real file may have different columns, so check the numbered list.

````
head -n 3 metadata.tsv
head -n 1 metadata.tsv | tr '\t' '\n' | cat -n
````

**Expected output** (example):
```
$ head -n 3 metadata.tsv
strain	virus	gisaid_epi_isl	genbank_accession	date	region	country	division
hCoV-19/Singapore/SGP-001/2024	betacoronavirus	EPI_ISL_9000001	?	2024-02-11	Asia	Singapore	Singapore
hCoV-19/Singapore/SGP-002/2024	betacoronavirus	EPI_ISL_9000002	?	2024-02-12	Asia	Singapore	Singapore
$ head -n 1 metadata.tsv | tr '\t' '\n' | cat -n
     1	strain
     2	virus
     3	gisaid_epi_isl
     4	genbank_accession
     5	date
     6	region
     7	country
     8	division
```

In this example `country` is column 7. Replace `7` with the column you want.

`cut -f` picks a column, `tail -n +2` skips the header row, `sort` groups equal values, `uniq -c` counts them, and `sort -nr` ranks them.

````
cut -f 7 metadata.tsv | tail -n +2 | sort | uniq -c | sort -nr
````

**Expected output** (example):
```
$ cut -f 7 metadata.tsv | tail -n +2 | sort | uniq -c | sort -nr
      4 Singapore
      3 Indonesia
      2 Thailand
      2 Malaysia
      1 Vietnam
```

## 10. Tools and the help habit
An environment is a separate toolbox. After activating `phylo`, your prompt shows `(phylo)`.

````
conda activate phylo
seqkit stats all_seqs.fasta
seqkit seq -n all_seqs.fasta | head -n 3
csvtk headers -t metadata.tsv
seqkit --help | head
conda deactivate
````

`seqkit stats` prints a small table (file, format, type, number of sequences, total length and more). Compare its sequence count with your `grep -c` result.

## 12. Cheat sheet
| Goal | Commands |
|---|---|
| Where am I / what is here | `pwd` `ls` `ls -lh` |
| Move around | `cd folder` `cd ..` `cd ~` `cd -` |
| Make / copy / move / delete | `mkdir` `cp` `mv` `rm` |
| Look inside | `head` `tail` `less` `wc -l` `zcat` |
| Search | `grep` `grep -c` |
| Save and chain | `>` `>>` `\|` |
| Tables | `cut -f` `sort` `uniq -c` |
| Help | `command --help` `seqkit -h` |

## 13. Common messages
| Message | Usually means |
|---|---|
| `No such file or directory` | typo, or you are in the wrong folder: check with `pwd` and `ls` |
| `command not found` | typo, or the right environment is not active |
| `Permission denied` | you are not allowed to change that file or folder |
| No `(base)` or `(phylo)` in the prompt | reopen the terminal, or run `conda activate` |
