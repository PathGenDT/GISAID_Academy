**Tips**

# 1. Prepare Windows
## 1.1 Turn on Windows features
1. Search **Turn Windows features on or off**
2. Tick **Windows Subsystem for Linux** and **Virtual Machine Platform**
3. Click OK and **restart the PC**

# 2. Install WSL (Ubuntu)
## 2.1 Install
1. Click Start and search **Windows PowerShell**
2. Right-click it and choose **Run as administrator**
3. Paste the command below and wait until it finishes

```
wsl --install
```

4. Close any pop-up
5. Create your Ubuntu account: type a **username**, press Enter
6. Type a **password**. Characters will not be displayed. Press Enter, then re-type it and press Enter
7. If asked to continue, type `y` or `n` and press Enter
8. **Restart the machine** when installation is complete

> **Write down your username and password.** You need the password for `sudo`.

## 2.2 Check that it worked
In PowerShell:

```
wsl -l -v
```

Example output. Ubuntu should show **VERSION 2**:
```
  NAME      STATE           VERSION
* Ubuntu    Running         2
```
## 2.3 Limit memory (only if your RAM is less than 16 GB)
This step is **mandatory for computers with less than 16 GB of RAM**. Skip it otherwise.
1. Press Win+R (the **Run** box) and type `%USERPROFILE%`
2. Create a new text document and paste the lines below
3. Click File > Save as, name it `.wslconfig`, and set *Save as type* to **All files**
4. Make sure the name is not `.wslconfig.txt`
```
[wsl2]
memory=8GB
swap=4GB
```
5. Apply it by running this in PowerShell, then open Ubuntu again

```
wsl --shutdown
```

# 3. Install Miniforge
Open **Ubuntu** from the Start menu and log in if asked. Your prompt should end with `$`.

## 3.1 Download and run the installer
`\` at the end of a line means the command continues on the next line. Paste both lines together.

```
cd ~
wget https://github.com/conda-forge/miniforge/releases/latest/download/\
Miniforge3-Linux-x86_64.sh
```

```
bash Miniforge3-Linux-x86_64.sh
```

- Follow the prompts (press Enter or Space to page through the licence)
- Type **yes** when asked to initialise conda
- **Close Ubuntu and open it again** so the change applies

## 3.2 Check it worked
Your prompt should now start with **(base)**. Then run:

```
conda --version
conda config --show channels
```

Expected: only **conda-forge** is listed.
```
channels:
  - conda-forge
```
If `(base)` is missing, run:

```
conda activate
```

> **Do not install tools in (base).** Create a new environment for each project.

Miniforge replaces the Miniconda installer used in earlier years. It defaults to conda-forge, so everyone gets the same packages.

# 4. Troubleshooting
| Problem | What to try |
|---|---|
| `wsl --install` is not recognised or fails | Update Windows, make sure PowerShell is run as administrator, and check that virtualisation is enabled in the BIOS |
| Computer slows down or WSL crashes | Apply the memory limit in section 2.3 |
| No `(base)` in the prompt | Close and reopen Ubuntu, or run `conda activate` |
| Channels list shows more than conda-forge | Tell an instructor before creating the environment |
| `Permission denied` or `sudo` fails | Check you are using the right password. See 7.1 if you are logged in as root |
| Solver error or `package not found` | Tell an on-site instructor. They will give an alternative method |
| Commands are slow in `/mnt/c/...` | Work in your Linux home (`~`), not the Windows drive |

## 4.1 Ubuntu opens as root (no user created)
In **Ubuntu**, replace `<username>` with the name you want:

```
adduser <username>
usermod -aG sudo <username>
```

Follow the prompts for `adduser`, then close Ubuntu. Open **PowerShell as administrator** and run:

```
wsl --manage Ubuntu --set-default-user <username>
```

Close PowerShell and open Ubuntu again to confirm the change. Your prompt should now show your username, not `root`.
