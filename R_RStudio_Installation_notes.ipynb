# Install R and RStudio
**GISAID Academy · Pre-workshop setup**

R is used for metadata cleaning, processing and visualisation. **R** is the programming language and **RStudio** is the friendly window for using it. You need both.

These are *notes*, not a running lab. Do the steps on your own computer and paste the code cells into the **RStudio Console**.

> Install R and RStudio on **Windows** (not inside Ubuntu/WSL). Install **R first**, then RStudio.

# 1. Install R
1. Go to https://cran.r-project.org/
2. Select the R download for your operating system
3. Run the downloaded installer (the `.exe` file on Windows)
4. Follow the installation wizard and keep the **default settings**
5. Click **Finish** when it completes

# 2. Install RStudio
1. Go to https://docs.posit.co/ide/user/download.html
2. Choose the free, open-source **RStudio Desktop** for your operating system
3. Download and run the installer (the `.exe` file on Windows)
4. Follow the on-screen instructions
5. Launch **RStudio** from the Start menu

New to RStudio? Read the getting-started guide before the course: https://docs.posit.co/ide/user/ide/get-started/

# 3. Install the R packages
Open RStudio. The **Console** is the window with the `>` prompt. Copy a cell below, paste it into the Console and press **Enter**.

## 3.1 Main packages

```
install.packages(c(
  "tidyverse",
  "readxl",
  "ape",
  "patchwork",
  "BiocManager"
))
```

## 3.2 ggtree (from Bioconductor)

```
BiocManager::install("ggtree", ask = FALSE)
```

## 3.3 maps

```
install.packages("maps")
```

Installation can take several minutes. Wait until the `>` prompt returns before pasting the next cell.

# 4. Check that it worked

```
library(tidyverse)
library(readxl)
library(ape)
library(patchwork)
library(ggtree)
library(maps)  # optional
```

If the packages load **without error messages**, the installation was successful. Messages such as *Attaching core tidyverse packages* are normal.

> Packages only need to be installed **once**, but `library()` must be run **each time** you start a new R session.

## What each package is for
| Package | Main use |
|---|---|
| tidyverse | Data manipulation and visualization (including ggplot2 and dplyr) |
| readxl | Importing data from Excel spreadsheets |
| ape | Reading, manipulating and analysing phylogenetic trees |
| patchwork | Combining multiple plots into a single figure |
| BiocManager | Installing packages from Bioconductor |
| ggtree | Drawing trees with metadata (installed from Bioconductor) |
| maps | Bubble map visualization |
